# Darwin-Evolab: Grand Multi-Stage Heavy-Compute Benchmark & Calibration Protocol
### Autonomous Multi-Domain Evolution across Software, Silicon, High-Dimensional Mathematics, and Interoceptive DNA Reading

**Core Objective**:
Harness Kaggle's extensive hardware capabilities (Multi-Core vCPUs + 30 GB RAM + up to 12-hour execution limit) to execute a comprehensive, 4-stage empirical benchmark of Darwin-Evolab without any LLM dependencies:

1. **Phase 1: Software Repair Track**: Run parallel multi-core AST search across the **full 300-instance SWE-bench Lite suite**, followed by `IntronDetector` backward ablation to isolate and prune hitchhiking code lines (400+ introns).
2. **Phase 2: Silicon Synthesis Track**: Evolve CMOS logic netlists (Half Adder, Full Adder, 2-bit Multiplier, 2-bit Comparator, Parity, and 3-bit Multiplier) under `NSGA2Engine` multi-objective optimization (Area vs Delay vs Dynamic Switching Power), verified against truth tables and analyzed via Julian Miller neutrality mapping.
3. **Phase 3: High-Dimensional Continuous Track**: Stress-test evolutionary scaling across 200D and 500D non-convex rugged landscapes (Rastrigin, Ackley, Rosenbrock, Griewank) with vectorized evaluation (over 1.2 Million evaluations), Turner Bloat Ratio auditing ($B(g)$), and Holland Building Block schema mining.
4. **Phase 4: Holistic Meta-Introspection & Strategic Calibration**: Run `DNAReader` and `EvolutionaryHistorian` to audit statistical Governor drift, enforce $\alpha = 0.05$ and $\epsilon = 10^{-6}$, run a 25,000-sample Dirichlet Dream-RSI replay optimization, and bundle all partitioned artifacts into a single downloadable zip.

**Zero-LLM Invariant**: 100% sovereign, local, symbolic AST transformations, Directed Acyclic Graph (DAG) netlists, and vectorized linear algebra.

```python
# Cell 1: Environment Diagnostics & Universal Workspace Setup
import os
import sys
import shutil
import zipfile
import multiprocessing
from pathlib import Path

print("=" * 80)
print(" DARWIN-EVOLAB: ENVIRONMENT & HARDWARE DIAGNOSTICS")
print("=" * 80)
cpu_count = multiprocessing.cpu_count()
print(f"[INFO] Logical CPU Cores Available: {cpu_count}")

try:
    import torch
    gpu_avail = torch.cuda.is_available()
    print(f"[INFO] PyTorch CUDA Available: {gpu_avail}")
    if gpu_avail:
        print(f"[INFO] Device Name: {torch.cuda.get_device_name(0)}")
except ImportError:
    pass

# Writable workspace in /kaggle/working
workspace_dir = Path("/kaggle/working/evolab_workspace")
workspace_dir.mkdir(parents=True, exist_ok=True)

print("\n[INFO] Scanning /kaggle/input for Darwin-Evolab dataset or code tree...")
input_root = Path("/kaggle/input")
found_sources = False

if input_root.exists():
    # 1. Look for any .zip archives in /kaggle/input and extract them to workspace
    zip_candidates = list(input_root.rglob("*.zip"))
    for z_path in zip_candidates:
        print(f"[INFO] Found archive at {z_path}. Extracting to {workspace_dir}...")
        with zipfile.ZipFile(z_path, "r") as zf:
            zf.extractall(workspace_dir)
        found_sources = True

    # 2. If no zip, Kaggle auto-extracts uploaded datasets into folders
    if not found_sources:
        for init_p in input_root.rglob("evolab/__init__.py"):
            dataset_dir = init_p.parent.parent
            if dataset_dir.name == "src":
                dataset_dir = dataset_dir.parent
            print(f"[INFO] Found uncompressed Darwin-Evolab at {dataset_dir}")
            print(f"[INFO] Copying to writable workspace {workspace_dir}...")
            for item in dataset_dir.iterdir():
                dest = workspace_dir / item.name
                if item.is_dir():
                    if dest.exists(): shutil.rmtree(dest)
                    shutil.copytree(item, dest)
                else:
                    shutil.copy2(item, dest)
            found_sources = True
            break

# 3. Fallback: check local working directory
if not found_sources:
    cwd = Path(".").resolve()
    for init_p in cwd.rglob("evolab/__init__.py"):
        src_p = init_p.parent.parent
        print(f"[INFO] Found evolab in local directory: {src_p}")
        if str(src_p) not in sys.path:
            sys.path.insert(0, str(src_p))
        found_sources = True

# Register standard candidate paths
for p in [
    workspace_dir / "src",
    workspace_dir,
    Path("/kaggle/working/src"),
    Path("/kaggle/working"),
    Path("."),
    Path("src"),
]:
    if p.exists() and str(p.resolve()) not in sys.path:
        sys.path.insert(0, str(p.resolve()))

# Deep scan across /kaggle to guarantee sys.path coverage
for init_path in Path("/kaggle").rglob("**/evolab/__init__.py"):
    pkg_root = str(init_path.parent.parent.resolve())
    if pkg_root not in sys.path:
        sys.path.insert(0, pkg_root)

try:
    import evolab
    print(f"\n[SUCCESS] Loaded evolab package v{evolab.__version__} from {evolab.__file__}")
except ModuleNotFoundError:
    print("\n[ERROR] Could not import 'evolab'.")
    print("Current sys.path:")
    for sp in sys.path:
        print(f"  - {sp}")
    if input_root.exists():
        print("\nAll items in /kaggle/input:")
        for f in input_root.rglob("*"):
            print(f"  {f}")
    raise
```

## Cell 2: Launch Industrial-Grade Multi-Core Benchmark with Live Popen Streaming
Executes the full industrial benchmark across all 4 vCPUs with live line-by-line log output:
- **Phase 1**: Full 300 instances of SWE-bench Lite with multi-core parallel processing (`ProcessPoolExecutor`).
- **Phase 2**: 6 Silicon CMOS Netlists (including 3-bit Multiplier and Parity) evolved over 150 generations under NSGA-II.
- **Phase 3**: 4 Continuous non-convex spaces (200D to 500D) evaluated over 500 generations (>1,200,000 evaluations).
- **Phase 4**: 25,000-sample Dirichlet Dream-RSI simulation with $\alpha = 0.05$ and $\epsilon = 10^{-6}$ Governor drift guard.
- **Runtime Expectation**: ~15 to 25 minutes of continuous 100% vCPU compute.

```python
# Cell 2: Launching the Industrial Grand Benchmark with Real-Time Popen Streaming
import subprocess
import sys
from pathlib import Path

script_path = None
for cand_script in [
    workspace_dir / "scripts" / "kaggle_grand_run.py",
    Path("scripts/kaggle_grand_run.py"),
    Path("/kaggle/working/scripts/kaggle_grand_run.py"),
]:
    if cand_script.exists():
        script_path = cand_script
        break

if not script_path:
    for s in Path("/kaggle").rglob("**/kaggle_grand_run.py"):
        script_path = s
        break

if not script_path:
    raise FileNotFoundError("kaggle_grand_run.py orchestrator not found!")

output_dir = "/kaggle/working/grand_run_output"

print(f"[INFO] Found orchestrator at {script_path}")
print(f"[INFO] Executing Full Industrial Benchmark across all vCPUs (Output: {output_dir})...")
print("[INFO] Live stdout streaming enabled via Popen (Expected runtime: ~15 to 25 minutes of 100% CPU crunching).\n")

cmd = [
    sys.executable, str(script_path),
    "--output-dir", output_dir,
    "--swe-instances", "300",
    "--cgp-gens", "150",
    "--num-gens", "500",
    "--enable-dna-reader",       # لتفعيل الفحص الشامل ومضاعفة رصد الإنترونات 400+
    "--governor-alpha", "0.05",  # لإجبار الحاكم على دلالة إحصائية صارمة p < 0.05
    "--governor-epsilon", "1e-6" # لمنع احتساب الانجراف المحايد كتحسين وهمي
]

# Real-time unbuffered line streaming so you see progress immediately
proc = subprocess.Popen(
    cmd,
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    bufsize=1,
)

for line in iter(proc.stdout.readline, ""):
    sys.stdout.write(line)
    sys.stdout.flush()

proc.stdout.close()
return_code = proc.wait()
if return_code != 0:
    print(f"\n[ERROR] Benchmark process exited with code {return_code}")
else:
    print("\n[SUCCESS] Industrial Grand Benchmark completed with 100% success!")
```
